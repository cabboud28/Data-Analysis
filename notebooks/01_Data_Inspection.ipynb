```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the three sheets
file_path = "../data/Superstore.xls"

orders = pd.read_excel(file_path, sheet_name="Orders")
returns = pd.read_excel(file_path, sheet_name="Returns")
people = pd.read_excel(file_path, sheet_name="People")

# Inspect dataset dimensions
print("ORDERS:", orders.shape)
print("RETURNS:", returns.shape)
print("PEOPLE:", people.shape)

# Inspect columns
print("\nOrders columns:")
print(orders.columns.tolist())

# Inspect missing values
print("\nMissing values:")
print(orders.isnull().sum())

# Inspect data types
print("\nData types:")
print(orders.dtypes)

# Profit distribution
print("\nProfit statistics:")
print(orders["Profit"].describe())
```

```python

# DATASET QUALITY AUDIT

datasets = {
    "Orders": orders,
    "Returns": returns,
    "People": people
}

for name, df in datasets.items():
    print(f"\n{'='*45}")
    print(f"DATASET: {name}")
    print(f"{'='*45}")

    print("Dimensions:", df.shape)
    print("Duplicate rows:", df.duplicated().sum())
    print("Missing values:", df.isnull().sum().sum())

    print("\nMissing values by column:")
    missing = df.isnull().sum()
    print(missing[missing > 0])

# ORDERS-SPECIFIC ANALYSIS

print("\n--- ORDERS ANALYSIS ---")

print("Distinct orders:", orders["Order ID"].nunique())
print("Distinct customers:", orders["Customer ID"].nunique())
print("Distinct products:", orders["Product ID"].nunique())

print("\nOrder date range:")
print(orders["Order Date"].min(), "to", orders["Order Date"].max())

print("\nTotal sales:", orders["Sales"].sum())
print("Total profit:", orders["Profit"].sum())

print("\nProfit distribution:")
print("Profitable:", (orders["Profit"] > 0).sum())
print("Loss-making:", (orders["Profit"] < 0).sum())
print("Break-even:", (orders["Profit"] == 0).sum())

print("\nProfit statistics:")
print(orders["Profit"].describe())

# RETURNS ANALYSIS

print("\n--- RETURNS ANALYSIS ---")

print("Distinct returned orders:",
      returns["Order ID"].nunique())

print("Duplicate return Order IDs:",
      returns["Order ID"].duplicated().sum())
```

```python

# INVESTIGATING THE RETURNS DATASET

print("Returned values:")
print(returns["Returned"].value_counts(dropna=False))

print("\nMost repeated Order IDs:")
print(
    returns["Order ID"]
    .value_counts()
    .head(15)
)

print("\nNumber of unique Order IDs:",
      returns["Order ID"].nunique())

print("\nNumber of unique complete rows:",
      returns.drop_duplicates().shape[0])

# Check whether duplicate Order IDs have conflicting return statuses
status_check = returns.groupby("Order ID")["Returned"].nunique()

print("\nOrder IDs with conflicting statuses:",
      (status_check > 1).sum())

# Check whether all returned Order IDs exist in Orders
orders_ids = set(orders["Order ID"])
returns_ids = set(returns["Order ID"])

print("\nReturned IDs found in Orders:",
      len(returns_ids.intersection(orders_ids)))

print("\nReturned IDs not found in Orders:",
      len(returns_ids - orders_ids))
```

```python

# EXPLORATORY DATA ANALYSIS
# 1. Numerical variable investigation

numerical_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

print("NUMERICAL SUMMARY")
print(orders[numerical_columns].describe().T)

print("\nPROFITABILITY DISTRIBUTION")

profit_status = pd.Series({
    "Profitable": (orders["Profit"] > 0).sum(),
    "Loss-making": (orders["Profit"] < 0).sum(),
    "Break-even": (orders["Profit"] == 0).sum()
})

print(profit_status)

print("\nPROFITABILITY PERCENTAGES")
print((profit_status / len(orders) * 100).round(2))
```

```python

# INVESTIGATION 2: DISCOUNT AND PROFITABILITY

# A. Discount vs Category
discount_category = pd.crosstab(
    orders["Category"],
    orders["Discount_Level"],
    margins=True
)

print("TRANSACTIONS BY CATEGORY AND DISCOUNT")
print(discount_category)

# B. Check losses with zero discount
no_discount_losses = orders[
    (orders["Discount"] == 0) &
    (orders["Profit"] < 0)
]

print("\nLOSS-MAKING TRANSACTIONS WITH NO DISCOUNT:")
print("Count:", len(no_discount_losses))

# C. Inspect high-discount transactions
high_discount = orders[orders["Discount"] >= 0.30]

print("\nHIGH DISCOUNT TRANSACTIONS")
print("Total:", len(high_discount))
print("Loss-making:", (high_discount["Profit"] < 0).sum())
print("Profitable:", (high_discount["Profit"] > 0).sum())

# D. Profit outliers
print("\nPROFIT OUTLIER INVESTIGATION")

Q1 = orders["Profit"].quantile(0.25)
Q3 = orders["Profit"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = orders[
    (orders["Profit"] < lower_bound) |
    (orders["Profit"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of potential outliers:", len(outliers))

print("\nLargest losses:")
print(
    orders.nsmallest(10, "Profit")[
        ["Category", "Sub-Category", "Product Name",
         "Sales", "Discount", "Profit"]
    ].to_string(index=False)
)

print("\nLargest profits:")
print(
    orders.nlargest(10, "Profit")[
        ["Category", "Sub-Category", "Product Name",
         "Sales", "Discount", "Profit"]
    ].to_string(index=False)
)
```

```python

# PROFITABILITY BY PRODUCT CATEGORY

category_analysis = (
    orders.groupby("Category")
    .agg(
        Transactions=("Profit", "size"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Loss_Transactions=("Profit", lambda x: (x < 0).sum())
    )
)

category_analysis["Loss_Rate_%"] = (
    category_analysis["Loss_Transactions"] /
    category_analysis["Transactions"] * 100
)

print(category_analysis.round(2))
```

```python

# PROFITABILITY BY DISCOUNT LEVEL

orders["Discount_Level"] = pd.cut(
    orders["Discount"],
    bins=[-0.001, 0, 0.10, 0.20, 0.30, 0.50, 1],
    labels=[
        "No Discount",
        "Low (1-10%)",
        "Moderate (10-20%)",
        "High (20-30%)",
        "Very High (30-50%)",
        "Extreme (Above 50%)"
    ]
)

discount_analysis = (
    orders.groupby("Discount_Level", observed=False)
    .agg(
        Transactions=("Profit", "size"),
        Average_Discount=("Discount", "mean"),
        Average_Profit=("Profit", "mean"),
        Total_Profit=("Profit", "sum"),
        Loss_Transactions=("Profit", lambda x: (x < 0).sum())
    )
)

discount_analysis["Loss_Rate_%"] = (
    discount_analysis["Loss_Transactions"] /
    discount_analysis["Transactions"] * 100
)

print(discount_analysis.round(2))
```

```python

# INVESTIGATION 3: CATEGORY AND DISCOUNT INTERACTION

orders["Loss_Making"] = (orders["Profit"] < 0).astype(int)

category_discount_analysis = (
    orders.groupby(
        ["Category", "Discount_Level"],
        observed=False
    )
    .agg(
        Transactions=("Profit", "size"),
        Average_Profit=("Profit", "mean"),
        Total_Profit=("Profit", "sum"),
        Loss_Count=("Loss_Making", "sum")
    )
    .reset_index()
)

category_discount_analysis["Loss_Rate_%"] = (
    category_discount_analysis["Loss_Count"] /
    category_discount_analysis["Transactions"] * 100
)

print(
    category_discount_analysis
    .round(2)
    .to_string(index=False)
)
```

```python

# INVESTIGATION 4: PROFIT OUTLIERS

print("Potential outliers:", len(outliers))
print("Percentage of dataset:",
      round(len(outliers) / len(orders) * 100, 2), "%")

print("\nLargest losses by category:")

print(
    outliers.nsmallest(10, "Profit")[
        ["Category", "Sub-Category", "Sales",
         "Quantity", "Discount", "Profit"]
    ].to_string(index=False)
)

print("\nLargest profits by category:")

print(
    outliers.nlargest(10, "Profit")[
        ["Category", "Sub-Category", "Sales",
         "Quantity", "Discount", "Profit"]
    ].to_string(index=False)
)
```

```python

# PROFITABILITY BY REGION AND CUSTOMER SEGMENT

def profitability_summary(data, group_column):
    result = (
        data.groupby(group_column)
        .agg(
            Transactions=("Profit", "size"),
            Total_Sales=("Sales", "sum"),
            Total_Profit=("Profit", "sum"),
            Average_Profit=("Profit", "mean"),
            Loss_Count=("Profit", lambda x: (x < 0).sum())
        )
    )

    result["Loss_Rate_%"] = (
        result["Loss_Count"] / result["Transactions"] * 100
    )

    return result.sort_values("Total_Profit", ascending=False)


print("PROFITABILITY BY REGION")
print(profitability_summary(orders, "Region").round(2))

print("\nPROFITABILITY BY CUSTOMER SEGMENT")
print(profitability_summary(orders, "Segment").round(2))
```

```python

# PROFITABILITY BY SUB-CATEGORY

subcategory_analysis = (
    orders.groupby("Sub-Category")
    .agg(
        Transactions=("Profit", "size"),
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Profit=("Profit", "mean"),
        Loss_Count=("Profit", lambda x: (x < 0).sum())
    )
)

subcategory_analysis["Loss_Rate_%"] = (
    subcategory_analysis["Loss_Count"] /
    subcategory_analysis["Transactions"] * 100
)

print("MOST PROFITABLE SUB-CATEGORIES")
print(
    subcategory_analysis
    .sort_values("Total_Profit", ascending=False)
    .head(10).round(2)
)

print("\nLEAST PROFITABLE SUB-CATEGORIES")
print(
    subcategory_analysis
    .sort_values("Total_Profit")
    .head(10).round(2)
)
```

```python

# CORRELATION ANALYSIS

numerical_data = orders[
    ["Sales", "Quantity", "Discount", "Profit"]
]

correlation_matrix = numerical_data.corr(method="pearson")

print("CORRELATION MATRIX")
print(correlation_matrix.round(3))
```

```python

# REGIONAL PROFITABILITY INVESTIGATION

regional_analysis = (
    orders.groupby(["Region", "Category"])
    .agg(
        Transactions=("Profit", "size"),
        Average_Profit=("Profit", "mean"),
        Total_Profit=("Profit", "sum"),
        Loss_Count=("Profit", lambda x: (x < 0).sum())
    )
    .reset_index()
)

regional_analysis["Loss_Rate_%"] = (
    regional_analysis["Loss_Count"] /
    regional_analysis["Transactions"] * 100
)

print("REGIONAL PROFITABILITY BY CATEGORY")
print(regional_analysis.round(2).to_string(index=False))


# Investigate discount patterns by region

regional_discount = (
    orders.groupby(["Region", "Discount_Level"])
    .agg(
        Transactions=("Profit", "size"),
        Average_Discount=("Discount", "mean"),
        Average_Profit=("Profit", "mean"),
        Loss_Count=("Profit", lambda x: (x < 0).sum())
    )
    .reset_index()
)

regional_discount["Loss_Rate_%"] = (
    regional_discount["Loss_Count"] /
    regional_discount["Transactions"] * 100
)

print("\nREGIONAL PROFITABILITY BY DISCOUNT")
print(regional_discount.round(2).to_string(index=False))
```

```python

# SHIPPING DURATION

orders["Order Date"] = pd.to_datetime(orders["Order Date"])
orders["Ship Date"] = pd.to_datetime(orders["Ship Date"])

orders["Shipping_Days"] = (
    orders["Ship Date"] - orders["Order Date"]
).dt.days

shipping_analysis = (
    orders.groupby("Ship Mode")
    .agg(
        Transactions=("Profit", "size"),
        Average_Shipping_Days=("Shipping_Days", "mean"),
        Average_Profit=("Profit", "mean"),
        Total_Profit=("Profit", "sum"),
        Loss_Count=("Profit", lambda x: (x < 0).sum())
    )
)

shipping_analysis["Loss_Rate_%"] = (
    shipping_analysis["Loss_Count"] /
    shipping_analysis["Transactions"] * 100
)

print(shipping_analysis.round(2))
```

```python
from scipy.stats import chi2_contingency, kruskal, spearmanr
import numpy as np
```

```python

# STATISTICAL TESTING

orders["Loss_Making"] = (orders["Profit"] < 0).astype(int)

# 1. Region and profitability status
region_table = pd.crosstab(
    orders["Region"],
    orders["Loss_Making"]
)

chi2_region, p_region, dof, expected = chi2_contingency(region_table)

print("REGION VS LOSS-MAKING STATUS")
print("Chi-square:", chi2_region)
print("P-value:", p_region)

# 2. Category and profitability status
category_table = pd.crosstab(
    orders["Category"],
    orders["Loss_Making"]
)

chi2_category, p_category, dof, expected = chi2_contingency(category_table)

print("\nCATEGORY VS LOSS-MAKING STATUS")
print("Chi-square:", chi2_category)
print("P-value:", p_category)

# 3. Shipping mode and profitability status
shipping_table = pd.crosstab(
    orders["Ship Mode"],
    orders["Loss_Making"]
)

chi2_shipping, p_shipping, dof, expected = chi2_contingency(shipping_table)

print("\nSHIPPING MODE VS LOSS-MAKING STATUS")
print("Chi-square:", chi2_shipping)
print("P-value:", p_shipping)

# 4. Profit differences across regions
region_groups = [
    group["Profit"].values
    for _, group in orders.groupby("Region")
]

stat_region, p_profit_region = kruskal(*region_groups)

print("\nPROFIT DIFFERENCES ACROSS REGIONS")
print("Kruskal-Wallis statistic:", stat_region)
print("P-value:", p_profit_region)

# 5. Shipping duration and profit
correlation, p_shipping_profit = spearmanr(
    orders["Shipping_Days"],
    orders["Profit"]
)

print("\nSHIPPING DAYS VS PROFIT")
print("Spearman correlation:", correlation)
print("P-value:", p_shipping_profit)
```

```python

import statsmodels.formula.api as smf

# Create an exploratory regression model
# Profit is the dependent variable

profit_model = smf.ols(
    "Profit ~ Sales + Quantity + Discount + C(Category) + C(Region) + C(Segment) + C(Q('Ship Mode'))",
    data=orders
).fit()

print(profit_model.summary())
```
